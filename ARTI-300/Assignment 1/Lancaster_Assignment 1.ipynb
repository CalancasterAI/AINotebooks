```python
# create user defined function that accepts 3 inputs - a, b, operationtype.  Operationtype values can include 
# add, subtract or multiple.  Based on the value of operationtype displays a+b, a-b, or a*b. Call the function at 
# least 3 times with different inputs
```

```python
def sampleMath(a, b, operationtype):
    if operationtype not in ['add', 'subtract', 'multiply']:
        print("Function sampleMath accepts: add, subtract, or multiply")
    elif operationtype == 'add':
        return f"{a} + {b} = {a+b}"
    elif operationtype == 'subtract':
        return f"{a} - {b} = {a-b}"
    elif operationtype == 'multiply':
        return f"{a} * {b} = {a*b}"

print(sampleMath(2,5,'add'))
print(sampleMath(3,4,'subtract'))
print(sampleMath(5,6,'multiply'))
```

```python
# create a lambda function that accepts to inputs a, b and cals a*b-b. Call the function at least 3 times with
# different inputs
```

```python
LambdaFunction = lambda a, b: a*b-b

print(LambdaFunction(4,5))
print(LambdaFunction(5,6))
print(LambdaFunction(6,7))
```
