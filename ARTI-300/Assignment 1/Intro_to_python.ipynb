### Collections of variables
There are 4 main types of collections of variables in python
- list - [ ] - can be of differnt types, ordered, changeable, allows duplicates, indexed by 0
- tuple - ( ) - can't change values
- set - { } - unordered and can't have duplicates
- dictionary - { } - unordered and indexed by strings


```python
list1 = ['ARTI300',6,12.3,'73']
print(list1)
print(list1[0])
print(type(list1[3]))
print(len(list1))
print(list1[1:3]) #first number is inclusive, second is exclusive
```

```python
# changing list values
list1[2] = 24.5
list1.append('added value') # adding an item to the list
print(list1)
list1.pop(1) # removing an item from the list
print(list1)
```

```python
# copying lists
list2 = list1
print(list2)
print(list1)
list2.append(14)
print(list2)
print(list1)
```

```python
# we must use the copy method to prevent this 
list3 = list1.copy()
print(list3)
print(list1)
list3[4] = 41
print(list3)
print(list1)
```

```python
tuple1 = (12,42,11,99,2351)
print(tuple1)
print(tuple1[2])
```

```python
# it is not possible to change entries in a touple
tuple1[2] = 5
```

### user defined functions

```python
# define a function that takes an input, squares it, adds 7, then returns the answer

def ConvertStringToUpper(s):
    """    documentation string example
    """
    ToUpper = s.upper()
    return ToUpper

print(ConvertStringToUpper("Welcome to ARTI300"))
print(ConvertStringToUpper.__doc__)
```

```python
ConvertStringToLower = lambda s: s.lower()
print(ConvertStringToLower("Welcome to ARTI300"))
```
