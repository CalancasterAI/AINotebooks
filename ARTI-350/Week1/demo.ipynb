```python
df[['Make', 'Model']] = df['Car name'].str.split(" ", n=1, expand=True)
```

```python
df = df.drop('Car name', axis=1)
df['Origin'] = df['Origin'].replace({1: 'america', 2: 'europe', 3: 'asia'})
df.head()
```

```python
df = pd.get_dummies(df, columns=['Origin'])
df.head()
```
