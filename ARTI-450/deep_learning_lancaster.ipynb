```python
import pandas as pd
data = pd.read_csv("/content/drive/My Drive/ColabFiles/ARTI-450/auto-mpg.csv", na_values='?', usecols=["Horsepower","MPG"])
data = data.dropna()
data.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(data['Horsepower'],data['MPG'], label='Data')
plt.xlabel('Horsepower')
plt.ylabel('MPG')
plt.xlim(0, 250)
plt.ylim(0, 50)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data['Horsepower'], data['MPG'], test_size=0.3, random_state=42)
```

<a name="no"></a>
## Training (0 hidden layers)


```python
from tensorflow.keras.layers import Dense, Input, InputLayer
from tensorflow.keras import layers

model0 = tf.keras.Sequential()
model0.add(layers.BatchNormalization())
model0.add(Dense(1, input_shape=(1,)))
```

```python
model0.compile(loss='mean_absolute_error', optimizer=tf.optimizers.Adam(learning_rate=0.1))
```

```python
model0.fit(X_train, y_train, epochs=100,verbose=0,validation_split = 0.2)
```

```python

x = tf.linspace(0.0, 250, 251)
y = model0.predict(x)

plt.scatter(X_train, y_train, label='Data')
plt.plot(x, y, color='k', label='Predictions')
plt.xlabel('Horsepower')
plt.ylabel('MPG')
plt.legend()
```

```python
# Accuracy (Root Mean Squared Error)
y_pred = model0.predict(X_test.values)
from sklearn.metrics import mean_squared_error
import numpy as np
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(rmse)
```

## Optimization

Slight learning rate tweak

```python
import tensorflow.keras as keras

model2 = keras.Sequential([
    layers.Input(shape=(1,)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model2.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.05),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
)

model2.fit(X_train, y_train,
          epochs=200,
          verbose=0,
          validation_split=0.2)
```

```python
# Accuracy (Root Mean Squared Error)
y_pred = model2.predict(X_test.values)
from sklearn.metrics import mean_squared_error
import numpy as np
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(rmse)
```

Polynomial flexibility

```python
data["Horsepower"] = pd.to_numeric(data["Horsepower"], errors="coerce")
data["MPG"] = pd.to_numeric(data["MPG"], errors="coerce")

X = data[["Horsepower"]].to_numpy(dtype=np.float32)
y = data[["MPG"]].to_numpy(dtype=np.float32)
```

```python
from tensorflow.keras import layers, regularizers

X2 = np.concatenate([X, X**2], axis=1)

model3 = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model3.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
    )

history = model3.fit(X2, y, epochs=200, verbose=0, validation_split=0.2)
```

```python
rmse = model3.evaluate(X2, y, verbose=0)[1]
print(rmse)
```

### Adding more features

I'm not sure if it's against the spirit of the assignment to add more features, but I honestly think that ~4.0 to 4.5 is the limit for what the model can predict with such a small single-feature dataset.

```python
features = ["Horsepower", "Weight", "Displacement", "Acceleration", "Model year"]
target = "MPG"

df = pd.read_csv("/content/drive/My Drive/ColabFiles/ARTI-450/auto-mpg.csv", na_values='?')

for c in features + [target]:
  df[c] = pd.to_numeric(df[c], errors="coerce")

df = df.dropna(subset=features + [target]).copy()

X5 = df[features].to_numpy(dtype=np.float32)
y = df[target].to_numpy(dtype=np.float32)
```

```python
model5 = keras.Sequential([
    layers.Input(shape=(X5.shape[1],)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model5.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
    )

history = model5.fit(X5, y, epochs=300, verbose=0, validation_split=0.2)
```

```python
rmse = model5.evaluate(X5, y, verbose=0)[1]
print(rmse)
```

Polynomial version

```python
df["hp2"] = df["Horsepower"] ** 2
df["hp_wt"] = df["Horsepower"] * df["Weight"]
features2 = features + ["hp2", "hp_wt"]
X6 = df[features2].to_numpy(np.float32)
```

```python
model6 = keras.Sequential([
    layers.Input(shape=(X6.shape[1],)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model6.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
    )

history = model6.fit(X6, y, epochs=300, verbose=0, validation_split=0.2)
```

```python
rmse = model6.evaluate(X6, y, verbose=0)[1]
print(rmse)
```

Adjusting batch size and epochs

```python
model6_batch = keras.Sequential([
    layers.Input(shape=(X6.shape[1],)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model6_batch.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
    )

history = model6_batch.fit(X6, y, epochs=300, verbose=0, batch_size=64, validation_split=0.2)
```

```python
rmse = model6_batch.evaluate(X6, y, verbose=0)[1]
print(rmse)
```

```python
model6_epoch = keras.Sequential([
    layers.Input(shape=(X6.shape[1],)),
    layers.BatchNormalization(),
    layers.Dense(1)
])

model6_epoch.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mse",
    metrics=[keras.metrics.RootMeanSquaredError()]
)

history = model6_epoch.fit(X6, y, epochs=900, verbose=0, batch_size=128, validation_split=0.2)
```

```python
rmse = model6_epoch.evaluate(X6, y, verbose=0)[1]
print(rmse)
```
